```python
numbers = [10, 15, 20, 25, 30, 35, 40, 45, 50, 55]

print("Even numbers:")
for n in numbers:
    if n % 2 == 0:
        print(n)
```

```python
students = {
    "Aishwarya": 85,
    "Rahul": 78,
    "Sneha": 92,
    "Arjun": 88,
    "Priya": 75
}

print("Student Marks:")

for name, marks in students.items():
    print(name, ":", marks)
```

```python
text = "python"

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print("Character frequencies:", frequency)
```
